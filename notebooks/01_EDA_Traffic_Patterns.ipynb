# Traffic Pattern EDA

Exploratory analysis of simulated IoT traffic sensor data across 5 intersections.

```python
import sys
sys.path.append("../src")
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from data_generator import generate_traffic_data

df = generate_traffic_data()
df.head()
```

## Traffic volume by hour of day

```python
hourly = df.groupby("hour")["vehicle_count"].mean()
plt.figure(figsize=(10,4))
hourly.plot(kind="bar", color="#1f77b4")
plt.title("Average Vehicle Count by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Avg Vehicle Count")
plt.tight_layout()
plt.show()
```

## Traffic state distribution

```python
df["traffic_state"].value_counts().plot(kind="pie", autopct="%1.1f%%", figsize=(6,6))
plt.title("Traffic State Distribution")
plt.ylabel("")
plt.show()
```

## Speed vs vehicle count relationship

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="vehicle_count", y="avg_speed_kmh", hue="traffic_state")
plt.title("Vehicle Count vs Average Speed")
plt.show()
```
